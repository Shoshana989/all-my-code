```python
from astropy.table import Table
from collections import defaultdict
import numpy as np

# Load catalog
catalog = Table.read('acc_ned_matches_table.txt', format='ascii.fixed_width')

# Build ned_name -> row indices
name_to_rows = defaultdict(list)
for i, name in enumerate(catalog['ned_name']):
    name_to_rows[name].append(i)

# Separate unique and repetitive groups
unique_names = {name for name, rows in name_to_rows.items() if len(rows) == 1}
repetitive_groups = [rows for rows in name_to_rows.values() if len(rows) > 1]
repetitive_names = {catalog[rows[0]]['ned_name'] for rows in repetitive_groups}

# Helper function
def has_redshift(row):
    val = row['redshift']
    try:
        return not np.isnan(float(val))
    except:
        return False


# Organize counters
def analyze(rows):
    summary = {
        'WISE': {'G': {'count': 0, 'with_z': 0}, 'IrS': {'count': 0, 'with_z': 0}},
        'DESI': {'G': {'count': 0, 'with_z': 0}, 'IrS': {'count': 0, 'with_z': 0}}
    }
    for row in rows:
        name = row['ned_name']
        typ = row['ned_type']
        if typ not in ('G', 'IrS'):
            continue
        if name.startswith('WISE') or name.startswith('WISEA'):
            key = 'WISE'
        elif name.startswith('DESI'):
            key = 'DESI'
        else:
            continue  # Skip if not WISE or DESI

        summary[key][typ]['count'] += 1
        if has_redshift(row):
            summary[key][typ]['with_z'] += 1
    return summary

# Apply analysis
unique_rows = [row for row in catalog if row['ned_name'] in unique_names]
repetitive_rows = [row for row in catalog if row['ned_name'] in repetitive_names]

unique_summary = analyze(unique_rows)
repetitive_summary = analyze(repetitive_rows)

# Pretty print
def print_summary(title, summary):
    print(f"\n{title}")
    for survey in ['WISE', 'DESI']:
        for typ in ['G', 'IrS']:
            count = summary[survey][typ]['count']
            with_z = summary[survey][typ]['with_z']
            print(f"{survey} {typ}: {count} total, {with_z} with redshift")

# Output results
print_summary("UNIQUE SOURCES", unique_summary)
print_summary("REPETITIVE SOURCES", repetitive_summary)
```

```python
from astropy.table import Table
from collections import defaultdict
import numpy as np

# Load catalog
catalog = Table.read('acc_ned_matches_table.txt', format='ascii.fixed_width')

# Build ned_name → row indices
name_to_rows = defaultdict(list)
for i, name in enumerate(catalog['ned_name']):
    name_to_rows[name].append(i)

# Identify repetitive groups
repetitive_groups = {name: rows for name, rows in name_to_rows.items() if len(rows) > 1}

# Redshift checker (robust)
def has_redshift_value(value):
    try:
        return not np.isnan(float(value))
    except:
        return False

# Analyze and print each group
print(f"{'Group Name':45} | {'Type':4} | {'Source':5} | {'Has Redshift':13} | Row indices")
print("-" * 100)
for name, rows in repetitive_groups.items():
    source_type = catalog[rows[0]]['ned_type']
    redshifts = [catalog[i]['redshift'] for i in rows]
    has_z = any(has_redshift_value(z) for z in redshifts)
    source_label = "WISE" if name.startswith("WISE") or name.startswith("WISEA") else "DESI" if name.startswith("DESI") else "OTHER"
    print(f"{name:45} | {source_type:4} | {source_label:5} | {'Yes' if has_z else 'No ':13} | {rows}")
```

```python
from astropy.table import Table
from collections import defaultdict
import numpy as np

# Load catalog
catalog = Table.read('acc_ned_matches_table.txt', format='ascii.fixed_width')

# Build ned_name → list of row indices
name_to_rows = defaultdict(list)
for i, name in enumerate(catalog['ned_name']):
    name_to_rows[name].append(i)

# Redshift checker
def has_redshift(value):
    try:
        return not np.isnan(float(value))
    except:
        return False

# Analysis containers
group_stats = {
    'repetitive': defaultdict(lambda: defaultdict(lambda: {'total': 0, 'with_z': 0})),
    'unique': defaultdict(lambda: defaultdict(lambda: {'total': 0, 'with_z': 0}))
}

# Print headers
print(f"{'GROUP TYPE':10} | {'ned_name':40} | {'Type':4} | {'Source':5} | {'Has Redshift':13} | Row indices")
print("-" * 110)

for name, rows in name_to_rows.items():
    group_type = 'repetitive' if len(rows) > 1 else 'unique'
    first_row = catalog[rows[0]]
    ned_type = first_row['ned_type']
    source = 'WISE' if name.startswith('WISE') or name.startswith('WISEA') else 'DESI' if name.startswith('DESI') else 'OTHER'
    has_z = any(has_redshift(catalog[i]['redshift']) for i in rows)

    # Update summary stats
    group_stats[group_type][source][ned_type]['total'] += 1
    if has_z:
        group_stats[group_type][source][ned_type]['with_z'] += 1

    # Print row/group info
    print(f"{group_type:10} | {name:40} | {ned_type:4} | {source:5} | {'Yes' if has_z else 'No ':13} | {rows}")

# Final summary
print("\nSUMMARY STATISTICS:")
for group_type in ['repetitive', 'unique']:
    print(f"\n{group_type.upper()} SOURCES:")
    for source in group_stats[group_type]:
        for typ in group_stats[group_type][source]:
            total = group_stats[group_type][source][typ]['total']
            with_z = group_stats[group_type][source][typ]['with_z']
            print(f"{source:5} {typ:4}: {total} total, {with_z} with redshift")
```
