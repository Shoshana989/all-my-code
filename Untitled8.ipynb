```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Fit line to AGN points

slope, intercept, _, _, _ = stats.linregress(x_log, y_log)
y_pred = slope * x_log + intercept
residuals = y_log - y_pred
std_dev = np.std(residuals)

# Perpendicular unit vector to line
norm = np.sqrt(1 + slope**2)
dx_perp = -slope / norm
dy_perp = 1 / norm

# Generate central line points
x_vals_log = np.linspace(.15, 4, 100)#max(x_log), 100)
y_vals_log = slope * x_vals_log + intercept

# Shift along perpendicular vector to make top and bottom edges
x_top = x_vals_log + std_dev * dx_perp
y_top = y_vals_log + std_dev * dy_perp

x_bot = x_vals_log - std_dev * dx_perp
y_bot = y_vals_log - std_dev * dy_perp

# Make closed polygon
x_box_log = np.concatenate([x_top, x_bot[::-1]])
y_box_log = np.concatenate([y_top, y_bot[::-1]])

# Convert back to linear for log-log plot
x_box = 10**x_box_log
y_box = 10**y_box_log
```
